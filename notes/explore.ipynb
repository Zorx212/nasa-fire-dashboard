```python
import pandas as pd
from pathlib import Path

for f in sorted(Path("../data/raw/PSI_159").glob("*.csv")):
    df = pd.read_csv(f, sep=None, engine="python", encoding="utf-8-sig")
    df.columns = df.columns.str.strip().str.replace(r"\s+", " ", regex=True)
    df = df.dropna(how="all")
    print("=" * 60)
    print(f.name, "| filas reales:", len(df))
    print(list(df.columns))
    for c in df.select_dtypes("object").columns:
        if df[c].nunique() <= 12:
            print(f"\n{c}:", df[c].value_counts(dropna=False).to_dict())
```

```python
import sys
from pathlib import Path
sys.path.append("..")
from src.ingest import ingest

lines = []
for key, df in ingest(Path("../data/raw")).items():
    if key.startswith("PSI_159"):
        continue  # ya está normalizado
    lines.append("=" * 60)
    lines.append(f"{key} | filas: {len(df)}")
    for c in df.columns:
        if c == "_psi_id":
            continue
        s = df[c]
        nunique = s.nunique(dropna=True)
        if s.dtype == "object" and nunique <= 8:
            detail = f"valores: {s.value_counts(dropna=False).to_dict()}"
        elif s.dtype != "object":
            detail = f"min={s.min()} max={s.max()}"
        else:
            detail = f"ej: {s.dropna().iloc[0] if s.notna().any() else 'vacío'}"
        lines.append(f"  {c!r} [{s.dtype}] nulos={s.isna().sum()} | {detail}")

text = "\n".join(lines)
open("exploracion_psi.txt", "w").write(text)
print(text)
```
